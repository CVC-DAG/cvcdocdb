```python
from cvcdocdb import NetworkXGraph, Node, WeakNode

graph = NetworkXGraph()
doc = Node(pk={"doc": "DOC-001"}, main_label="Document")
section = WeakNode(parent=doc, pk={"section": 1}, main_label="Section")
graph.insertNode(doc, replace=True)
graph.insertNode(section, insert_parent=True)
print(graph.get_nodes())
```

## Explore the graph

`show()` opens the cvcdocdb graph viewer on this graph. Click a node to see its properties,
double-click it to expand its neighbours, and search or filter from the sidebar. The viewer reads
the graph live from the kernel, so the graph is only closed in the last cell, once you are done.

`load_all=True` draws every node and relation; the `HAS` relation links the section to its document.

```python
from cvcdocdb.viewer import show

show(graph, load_all=True)
```
